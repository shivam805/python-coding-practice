```python
for i in range(len(l1)):
  for j in range(i+1, len(l1)):
    if l1[i] > l1[j]:
      l1[i], l1[j] = l1[j], l1[i]

print(l1)
print(l1[::-1])
```

```python
#decorator
def my_decorator(func):
  def wrapper():
    print("Hi AI")
    func()
  return wrapper()

@my_decorator
def abc():
  print("Hi Python")
abc
```

```python
l1 = [5,8,1,3]
l2 = [i*2 for i in l1]
print(l2)
```

```python
g1 = (a*2 for a in l1)
print(next(g1))
print(next(g1))
print(next(g1))
```
